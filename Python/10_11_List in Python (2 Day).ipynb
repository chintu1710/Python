## List
## What is list and why we use it?
### List is collection of hetrogenous data type which is mutable and has indexing.
### If we want to store multiple values in a single variable and can be modify and access easily with their index then we have to create a list.

### Example : 
#### Suppose you want to store names of your classmates in normal approach you will create name variable and assign their name to it but in this condition the number of variable will be equal to number of classmates. It is too difficult to remember them.
#### But with the list capability you can store all names of your classmates in single variable and can acess with single name with indexing. 

### Properties of list
#### 1. Mutable --> can be modified after creation 
#### 2. Indexed --> We can acess the value by index
#### 3. Hetrogenous --> Can store multiple different data types
#### 4. Elements are seperated by ,

## Indexing and slicing in list
### syntax ---->  var[start point : end point -1 : step size]
### Indexing is acessed by the [ ].
### 1. Positive indexing starts from 0 left to right
### 2. Negative Indexing starts from -1 right to left

````python
# Positive Indexing
fruits = ['apple','mango', 'orange', 'pineapple', 'kiwi', 'chiku' ]
print(len(fruits))
print(fruits[0])
````

````python
print(fruits[5])
````

````python
print(fruits[-5])
````

````python
print(fruits[-1])
````

````python
# Slicing
fruits[1:5:2]
````

````python
fruits[1:5:-2]
````

````python
fruits[-1:-5:-2]
````

````python
# Reverse the list
print(fruits)
print(fruits[::-1])
````

## zip

````python
print(zip().__doc__)
````

````python
# I want to add element of two different list by their index item
````

````python
list1 = [1,2,3,4,5]
list2 = [100,200,300,400,500]
````

```
1 + 100 = 101
2 + 200 = 202
3 + 300 = 303 

````python
# Solution 1
for i in range(0,5):
    print(f'{list1[i]}+{list2[i]} = {list1[i]+list2[i]}')
````

````python
# Solution 2 with zip
for i,j,k in zip(list1, list2, range(10,60,10)):
    print(f'{i}+{j} = {i+j} , {k}')
````
