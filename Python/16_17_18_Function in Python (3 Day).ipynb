```python
def addition(*a):
    res = 0
    st = ""
    b = 0
    sum_collection_dtype =0
    for i in a:
        if type(i) == int or type(i) == float:
            res +=i
        elif type(i) == str:
            st=st+i
        elif type(i) == bool:
            b+=1
        else:
            for j in i:
                sum_collection_dtype+=j

    return {'num_sum':res, 'string':st, 'boolean':b, 'sum_num_collection': sum_collection_dtype}
```

```python
a = [2,3,5]
b = (100,200,100)
s = {5,9,8}
d = {1:1000, 2:2000}

addition(10,20,True,'Shalini',a,b,s,d)
```

```python
# 4. Arbitrary Keyword Arguments Functions
# Parameter define with **
# This type of function will declare when we don't know the how many parameter will be given by the user
# but We are sure that a minimum definite numbner of parameter will be given with their key value pair.
# it take the parameter as a dictionary
# Both argument and parameter given by the user but developer will aware of the name of the minium parameter
```

```python
def salutationName(**a):
    print(a)
    if a['gender'].lower() == 'male':
        print(f'Hello Mr. {a["name"]}')
    else:
        print(f'Hello Ms. {a["name"]}')

```

```python
salutationName(college = 'IPS', name = 'Kavita', gender ='Female')
```

```python
salutationName(gender ='Female', name="Garima", surname = 'Tayde')
```

```python
salutationName(gender ='male', name="Kapil", surname = 'sharma')
```
