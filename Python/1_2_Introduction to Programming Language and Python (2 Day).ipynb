```python
# ^^^^^^^^^^^^^^^^^^^ print ^^^^^^^^^^^^^^^^^^^^^^
# print --> It is a builtin function that print the output on th console
# in print function ---> sep and end parameter are given we can use them as per need
print(print.__doc__)
```

```python
print('Aishwarya') # Default value of end = '\n' that means new line
print('Karishma')
print('Kapil')
print('Virat')
```

```python
print('Aishwarya', end = "") # We can change the default value as per need.
print('Karishma',  end = "3333")
print('Kapil',  end = "$$$$$")
print('Virat')
```

```python
li = ['Aishwarya','Karishma', 'Kapil', 'Virat', 'Aditya']
for i in li:
    print(i)
```

```python
li = ['Aishwarya','Karina', 'Kapil', 'Virat', 'Aditya']
for i in li:
    print(i, end = "------")
```

```python
# sep ---> it seperate the two or more variable from each other defalut value = " "
print('Aishwarya',"Karina") # Default value of sep = ' '
```

```python
print('Aishwarya',"Karina", sep="+++") # seperate the two variable with +++ or you can specify any character
```

```python
print('Aishwarya',"Karina", sep="$$$$$")
```

```python
print('Aishwarya', sep="+++")
```
