## Try and Except
## Why we need the try and except?
When an error occurs, or exception as we call it, Python will normally stop and generate an error message.

These exceptions can be handled using the try statement:

The try and except statement in Python is used to handle errors. When an error occurs, the code in the except block is executed. This allows the program to continue running even if an error occurs.

## What is try and except ?
A Try-Except statement is a code block that allows your program to take alternative actions in case an error occurs. Python will first attempt to execute the code in the try statement (code block 1). If no exception occurs, the except statement is skipped and the execution of the try statement is finished.

## Finally Block
### I am independent from all other block
### This block of code run if there is error in try block and it also run if there is no error in try block

````python
# Ex
try:
    num = input('Enter the number without decimal to get double : ')
    print(f'{int(num)*2}')
except Exception as t:
    print(f'\t\t{t}')
else:
    print(f'You got the double of {num}')
finally:
    print('This is the code of finally block')
    num = float(num)
    if num > 20:
        print(f'{num} is greater than 20')
    else:
        print(f'{num} is less than 20')
# print(num*2)
````
