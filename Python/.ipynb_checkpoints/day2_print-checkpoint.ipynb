```python
def atm_validate(**a):
    pin = a['pin']
    cardno = a['card']
    cards = {12345 : 1010, 98765 : 9898, 55443:5050, 14256 : 1234 }
    balance = {12345 : 0, 98765 : 500000, 55443:10000, 14256 : 3000 }
    
    if cardno in cards and pin == cards[cardno]:
        return f'user : Valid, Balance: {balance[cardno]}'
    else:
        return f'user : Not Authorised.'
        
```

```python
atm_validate(pin = 1234, card = 14256, name = 'SBI')
```

```python
pin = int(input('Enter your pin :'))
card = int(input('Enter you card no : '))
atm_validate(pin = pin, card = card, name = 'SBI')
```
