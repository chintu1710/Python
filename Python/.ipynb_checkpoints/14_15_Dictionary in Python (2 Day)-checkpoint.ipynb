## dict
A dictionary in Python is a collection of key-value pairs. It is unordered,changeable and indexed*.

Indexed according to keys


```python
# Creation of dictionary
# Method 1
d = {'key1' : 'Value1', "A":'apple', 'yellow':'Mango', 'summer':"sunny day"}
print(type(d))
print(d)
```

```python
d[1] # Error because there no key as 1 and you can access value by its key name
```

```python
d['A']
```

```python
d['summer']
```

```python
# Method 2
li = [2,3,4]
d2 = dict(li)
print(d2)
```

```python
t = (2,3,4)
d2 = dict(t)
print(d2)
```

```python
s = {2,3,4}
d2 = dict(s)
print(d2)
```

```python
k = (('T','test'),)

d2 = dict(k)
print(d2)
```

```python
a,b = ('T','test')
print(a)
print(b)
```

```python
l = [['m','mango'], ['k','kite']]

d2 = dict(l)
print(d2)
```

## copy()

```python
print(d.copy.__doc__)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}

d1 = d # it is not a copy it is a reference sharing
print('d1 =', d1)
print('d =', d )
d1['m'] = 'Monkey'
print('After the changes in d1')

print('d1 =', d1)
print('d =', d )
print(id(d1))
print(id(d))
```

```python
# how to create a copy
d = {'m':'mango','k':'kite', 'i' : 'icecream'}

d1 = d.copy() # it is a shallow copy of d
print('d1 =', d1)
print('d =', d )
d1['m'] = 'Monkey'
print('After the changes in d1')

print('d1 =', d1)
print('d =', d )
print(id(d1))
print(id(d))
```

## pop()

```python
print(d.pop.__doc__)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
a = d.pop('a', 'Key is not found') # set the value of a ="Key is not found"
print('Deleted Value = ', a)
print(d)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
a = d.pop('a',) # error 
print('Deleted Value = ', a)
print(d)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
a = d.pop('m',) # No error Key : value pair of m will be removed 
print('Deleted Value = ', a)
print(d)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
a = d.pop() # error 
print('Deleted Value = ', a)
print(d)
```

## popitem()

```python
print(d.popitem.__doc__)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
a = d.popitem() # it remove the last key value pair if key not defined 
print('Deleted Value = ', a)
print(d)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
a = d.popitem('m') # error because is does not take any argument default last key value pair 
print('Deleted Value = ', a)
print(d)
```

```python
d = {}
a = d.popitem() # error 
print('Deleted Value = ', a)
print(d)
```

## items()

## How to insert a new key value pair in dict

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
print(d)
# Add new key value pair
print('After addition of new key value pair')
d['s'] = 'ship'
print(d)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
print(d)
# Add new key value pair
print('After addition of new key value pair')
d['m'] = 'master'
print(d)
```

## update()

```python
print(d.update.__doc__)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
print(d)
d.update({'a': 'angoor'})
print(d)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
print(d)
d.update((('a', 'angoor'), ('b', 'bangal')))
print(d)
```

```python
d = {'m':'mango','k':'kite', 'i' : 'icecream'}
print(d)
d.update((('a'), ('b')))
print(d)
```
