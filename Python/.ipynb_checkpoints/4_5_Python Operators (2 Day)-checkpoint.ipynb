```python
# ----------------------- 5. Membership operators----------------------
# 1. in --> It retruns True if the substring or element is present inside the iterable
# 2. not in
```

```python
# What are iterables?
# Basically these are the collection data type and can be iter their element by index or over the loop

# Defination 2 ----
# Iterable is an object which can be looped over or iterated over
# with the help of a for loop. Objects like lists, tuples, sets, dictionaries, strings
```

```python
# list -- Iterable
li = ['a', 'b', 'Shalini', 'Isha', 'c', 'd']
print('Isha' in li)

st = 'Aish is an actress in bollywood'
print('an' in st)

fruits = ('apple', 'mango', 'kiwi', 'orange', 'strawberry')
print('kiwi' in fruits)
```

```python
# not in --> It retruns True if value is not present in the iterable


# list -- Iterable
li = ['a', 'b', 'Shalini', 'Isha', 'c', 'd']
print('Isha' not in li) # False

st = 'Aish is an actress in bollywood'
print('Abhishek' not in st) # True -- > Abhishek is not present in the string
```

```python
#>>>>>>>>>>>>>>>>>>>>>>>>>>> 6. Identity operators <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
# 1.  is --> it returns True if the object refers the same refrence

a = 5+5
b = 5+3+2
print(a is b)
print("Reference of a =  ",id(a))# id -- it gives the reference of the value stored in variable
print("Reference of b =  ",id(b))
```

```python
a = 5
b = 3.0+2
print(a is b)
print("Reference of a =  ",id(a))# id -- it gives the reference of the value stored in variable
print("Reference of b =  ",id(b))
```

```python
# ............... Difference between is and == ................................
a = 5
b = 5.0
print("Result of a == b ", a == b) # It compares the value
print("Reference of a =  ",id(a))
print("Reference of b =  ",id(b))
print("Result of a is b ", a is b) # It compares the reference
```

```python
# --------------------- 7. Bitwise operators  ---------------------------
# Bitwise operator works on the bit position not on the values
# &, |, ~, >> 1,  << 1
#

```

```python
bin(4) # it converts the decimale value into binary
```

```python
bin(5)
```

```python
5 & 8
```

```python
# Bitwise & (and)
# & ---> All True then True ( all 1 then 1)

# Bit_value    bin (4)      bin(5)      & 
# 2^7 = 128    0             0          0
# 2^6 = 64     0             0          0
# 2^5 = 32     0             0          0
# 2^4 = 16     0             0          0
# 2^3 = 8      0             0          0
# 2^2 = 4      1             1          1
# 2^1 = 2      0             0          0
# 2^0 = 1      0             1          0
# Total        4             5          4
print(4 & 5)
```

```python
bin(7)
```

```python
# Bitwise | (or)
# | ---> All False then False and if any True then True ( if any 1 then 1)

# Bit_value    bin (8)      bin(7)      | 
# 2^7 = 128    0             0          0
# 2^6 = 64     0             0          0
# 2^5 = 32     0             0          0
# 2^4 = 16     0             0          0
# 2^3 = 8      1             0          1
# 2^2 = 4      0             1          1
# 2^1 = 2      0             1          1
# 2^0 = 1      0             1          1
# Total        8             7          15 
print(16 | 7)
```

```python
# Bitwise ~ (negation) (not)
# ~ ---> Inverse the bit value 0 -->1  or 1 --> 0

# Bit_value    bin (8)      ~       
# 2^7 = 128    0             1          
# 2^6 = 64     0             1          
# 2^5 = 32     0             1          
# 2^4 = 16     0             1          
# 2^3 = 8      1             0          
# 2^2 = 4      0             1          
# 2^1 = 2      0             1          
# 2^0 = 1      0             1          
# Total        8            -9          
print(~ 555555)
```

```python
# shift << 1  (left shift by 1)
# It shifts the bit position by given number upward simply multiply 2^number

# Bit_value    bin (8)      << 1       
# 2^7 = 128    0             0          
# 2^6 = 64     0             0          
# 2^5 = 32     0             0          
# 2^4 = 16     0             1          
# 2^3 = 8      1             0         
# 2^2 = 4      0             0          
# 2^1 = 2      0             0          
# 2^0 = 1      0             0          
# Total        8            16          
print(16 << 9)
```

```python
print(8 << 2)
```

```python
# shift >> 1  (right shift by 1)
# It shifts the bit position by given number downward simply divide 2^number

# Bit_value    bin (8)      >> 1       
# 2^7 = 128    0             0          
# 2^6 = 64     0             0          
# 2^5 = 32     0             0          
# 2^4 = 16     0             0          
# 2^3 = 8      1             0         
# 2^2 = 4      0             1          
# 2^1 = 2      0             0          
# 2^0 = 1      0             0          
# Total        8             4          
print(8 >> 1)
```

```python
print(8 >> 4)
```
