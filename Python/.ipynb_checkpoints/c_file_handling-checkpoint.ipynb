## File Handling

## What is file handling?

#### File handling is the process of reading, writing, and modifying data in files. It is an essential part of many Python programs, as it allows data to be stored and accessed persistently. Python provides a number of functions and modules for file handling, making it easy to work with files in your code.

## Helpful functions for file handling
### 1. Mode of create the file - 'w', 'x', 'w+', 'a', 'a+'
       write() --> to write the data in file
### 2. Mode of read the file - 'r', 'r+'
       read() --> to read the data from file
            parameter of reading function - read, readline, readlines
### 3. Some Cursor methods - tell(), seek()
### 4. close()
## Another method to create and read file with context manager
### in this method we use the with keyword to avoid the close fucntion of file

## When we use 'w' mode to create a file it create a file if not exist
## but if file exists it will overwrite the file.

```python
n = 5
for i in range(1,11):
    print(f'{n}x{i} = {n*i}')
```

# Read methods
# read,
# readline, 
# readlines


```python
# now we want to read the file in python
# so we have to use the r mode
f = open('chintu.txt', 'r')
data = f.read() # read all the data at once
print(data)
f.close()
```

```python
f = open('chintu.txt', 'r')
data = f.readline() # read only one line by line
print(data)
# data2 = f.readline()
# print(data2)
f.close()
```

```python
f = open('chintu.txt', 'r')
data = f.readlines() # read all line at once but it gives the result in list
print(data)
f.close()
```

```python
for i in data:
    print(i)
```

```python
# now try create a file in w mode and try to read it
a = open('test123.txt', 'w')
a.write('Hello text\n')
a.write('Bye text file\n')
print('Message : Writing process complete')
data = a.read()
print(data)
a.close()
```

```python
# now try create a file in w+ mode and try to read it
a = open('test123.txt', 'w+')
a.write('Hello text\n')
a.write('Bye text file\n')
print('Message : Writing process complete')
print(a.tell()) # it gives the cursor position in file
data = a.read()
print(data)
a.close()
```

```python
# now try create a file in w+ mode and try to read it
a = open('test123.txt', 'w+')
a.write('Hello text\n')
a.write('Bye text file\n')
print('Message : Writing process complete')
print(a.tell()) # it gives the cursor position in file
a.seek(0) # set the cursor position at given index
data = a.read()
print(data)
a.close()
```

```python
# now we want to add more data or text in existing file at the end of 
# previous text
# so we have to use the a mode or a+ mode to write and read at same time
b = open('chintu.txt', 'a+')
b.write('\nLine3 test data by "a" mode')
print('Write process complete')
print(b.tell()) # current cursor position
b.seek(0) # set cursor position at 0
data = b.read() # read all data at once
print(data)
b.close()
```
