#### What is string in python?
A string in Python is a
n immutable sequence of characters. Strings are represented in Python by the str type. Strings can be created by enclosing a sequence of characters in either single or double quotes or triple quotes.
It is a sequence data type. It has indexing.

### Indexing
Means we can access the the character or element by its position.
There are two types of indexing
1. Positive (starts from 0 to n-1 left to right )
2. Negative (starts -1 to -n right to left)

We can also do the slicing with index find the sub string by 

```python
# multiple value or word
# so we have write a slicing 
# syntax for slicing   [start point : endpoint - 1 : step size]
st = 'Any Body can Do everything'
a = st[0:3]
print(a)
b = st[9:12]
print(b)
c = st[9:15]
print(c)
```

```python
# slicing with stepsize
st = 'Any Body can Do everything'
a = st[0:26:2]
print(a)
```

```python
st = 'Any Body can Do everything'
a = st[0:26:3]
print(a)
```

```python
st = 'Any Body can Do everything'
a = st[-26:-1]
print(a)
```

```python
st = 'Any Body can Do everything'
a = st[-1]
print(a)
```

```python
st = 'Any Body can Do everything'
a = st[-1:-4]
print(a)
```

```python
st = 'Any Body can Do everything'
a = st[-4:-1]
print(a)
```

```python
st = 'Any Body can Do everything'
a = st[-4:]
print(a)
```

```python
st = 'Any Body can Do everything'
a = st[-10:-1:3]
print(a)
```

```python
st = 'chintu chaure'
a = st.capitalize()
print(a)
```

```python
st = 'chintu chaure'
print(len(st))
a = st.center(20)
print(a)
print(len(a))
a = st.center(20, '~')
print(a)
```
