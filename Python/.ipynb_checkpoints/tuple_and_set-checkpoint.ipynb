# tuples
## Tuple
#### Tuple is a collection of hetrogenous data type which immutable in nature. It has indexing.

```python
a = ('chintu','srajal', 'mohan', 'payal')
print(type(a))
```

```python
a = ['chintu','srajal', 'mohan', 'payal']
print(type(a))
```

```python
a = {'chintu','srajal', 'mohan', 'payal'}
print(type(a))
```

```python
a = {'c':'chintu','s': 'srajal','m': 'mohan', 'p':'payal'}
print(type(a))
```

```python
a = ('chintu','srajal', 'mohan', 'payal')
print(type(a))
print(a)
```

```python
# try to add a new element in tuple
a[4] = 'Vinay'
```

```python
del(a[0])
```

```python
# We can not change anything in tuple after creation so it is immutable
```

```python
#  Methods in typle
# 1. count
# 2. index
```

```python
a = ('chintu','srajal','chintu' , 'payal','mohan', 'chintu', 'payal', 'chintu')
```

```python
a.count('chintu')
```

```python
a.index('chintu')
```

```python
a.index('chintu',1)
```

```python
a.index('chintu',3)
```

```python
a.index('chintu',6)
```

```python
a = ('chintu','srajal','chintu' , 'payal','mohan', 'chintu', 'payal', 'chintu')
# find frequency of given value and all index position
w = 'chintu'
c = a.count(w)
idx = 0
for i in range(c):
    print(a.index(w,idx))
    idx = a.index(w,idx)+1
```

# Set ##########

```python
# Set is collection data type which stores only unique value. can store hetrogenous 
# data type
```

```python
s = {'a','b', 'c', 'd', 'a', 'a', 'b', 'c'}
print(s)
```

```python
a = {'a', 'b', 'c', 'd'}
b = {'x', 'y', 'c', 'd'}
r = a.intersection(b) # it returns the common value of from both sets
print(r)
```

```python
a = {'a', 'b', 'c', 'd'}
b = {'x', 'y', 'c', 'd'}
r = a.difference(b) # it gives the uncommon value of base set
print(r)
```

# unpacking of list and tuple

```python
li = ['apple', 'mango' ,'kiwi', "oran'ge"] # packing

# upacking   assign the single value to a single variable

a,b,c,d = li
print(a)
print(b)
print(c)
print(d)
```

```python
a,b,c,d,e = li
```

```python
a,b,c = li
```

```python
print(f'list = {li}')
a,b,*c = li
print(a)
print(b)
print(c)
```

```python
print(f'list = {li}')
a,*b = li
print(a)
print(b)
```
