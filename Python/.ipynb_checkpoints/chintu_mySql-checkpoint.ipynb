**What is database?**

A database is a structured and organized collection of data that is designed for efficient storage, retrieval, and management. It is a fundamental component of modern computing and is used to store, organize, and manipulate data for various purposes. Databases are commonly used in a wide range of applications and industries, including business, science, government, healthcare, and more.

Key characteristics of databases include:

1. **Data Structure** - tables or relations
2. **Data Integrity** - unique keys, foreign keys, and data validation rules
3. **Data Retrieval** - retrieve specific information using database query languages
4. **Data Security** -  access control and security features

5. **Data Scalability** -  Databases can handle large amounts of data and are designed to scale as data volume grows. Different database systems offer various scaling options, such as vertical scaling (adding more resources to a single server) or horizontal scaling (distributing data across multiple servers).

6. **Transaction Management**: Databases support transaction management, allowing multiple operations to be grouped together as a single unit. This ensures that data remains in a consistent state, even in the presence of concurrent access.

There are different types of databases, including:

1. **Relational Databases**: These databases use tables to store data and are based on the principles
of relational algebra. Examples include MySQL, PostgreSQL, Oracle Database, and
Microsoft SQL Server.

2. **NoSQL Databases**: NoSQL databases are designed to handle unstructured or semi-structured data
and offer more flexible data models. Types of NoSQL databases include document stores
(e.g., MongoDB), key-value stores (e.g., Redis), column-family stores
(e.g., Cassandra), and graph databases (e.g., Neo4j).

3. **NewSQL Databases**: These databases combine the benefits of both relational and NoSQL databases,
aiming to provide high scalability and maintain the transactional consistency of
relational databases.

4. **In-Memory Databases**: These databases store data in RAM for extremely fast data retrieval,
making them suitable for real-time applications. Examples include Redis and Apache Ignite.

What is MySQL
MySQL is an open-source relational database management system (RDBMS) that is widely used for managing structured data. It is one of the most popular database management systems in the world and is known for its performance, scalability, and ease of use. MySQL is often used in web applications and is the default database system for many web development platforms and content management systems.

```python
import mysql.connector as sql # use the alia sql short the name
```

```python
# Making connection with Mysql via Mysql.connector
try:
  con = sql.connect(
  host="localhost",
  user="root",
  password="root"
  )
except Exception as e:
  print(e)
else:
  print("Connection Done") # it show that connection is established
```

```python
# creating a cursor for fire the qurey

cursor = con.cursor() # creation of cursor

# Show existing databases in RDBMS
query = 'show databases' # normal sql query in string format

cursor.execute(query) # fire the query

cursor.fetchall() # getting the result of query
```

```python
cursor = con.cursor()
query = 'show databases'
cursor.execute(query)
for db in cursor: # Method 2 for check the data in query
    print(db)
```

```python
# Create databases in RDBMS
try:
    cursor = con.cursor()
    dbname = input('Enter the Database name => ')
    query = f"create database {dbname}"
    cursor.execute(query)
except Exception as e:
    print(e)
else:
    print('Database Created')
```

```python
# Select the required database

try:
    cursor = con.cursor()
    dbname = input('Enter the Database name => ')
    query = f"use {dbname}"
    cursor.execute(query)
except Exception as e:
    print(e)
else:
    print(f'Database {dbname} selected for  use')
```

```python
# print all the tables of database
try:
    cursor = con.cursor()
    query = "show tables"
    cursor.execute(query)
    if len(cursor.fetchall()) == 0:
        print('No table present database')
    else:
        cursor.execute(query)
        print(cursor.fetchall())
except Exception as e:
    print(e)
```

```python
# create static table in database
try:
    cursor = con.cursor()
    query = "create table user(idno int, name varchar(30), mobile varchar(10), email varchar(50))"
    cursor.execute(query)
except Exception as e:
    print(e)
else:
    print('Table Created')
```

```python
# create user entry base table in database
tbl_name = input('Enter the table name => ')
query = f'create table {tbl_name}('

while True:
    col_name = input("Enter the column name => ")
    query = query+col_name+" "
    col_dtype = input("Enter the datatype of column => ")
    if col_dtype == 'int':
        query = query+col_dtype+", "
    else:
        col_length = input("Enter the column length => ")
        query = query+col_dtype+"("+ col_length+")" + ","
    next_col = input('Do you want to add more column "Y" for Yes "N" for No =>')
    if next_col.upper() == "N":
        query = query+")"
        query = query.replace(",)", ")")
        break
print(query)
try:
    cursor = con.cursor() # creating a  cursor
    cursor.execute(query)
    con.commit()
except Exception as e:
    print(e)
else:
    print(f'{tbl_name} is created')
```

```python
# To enter the data into table in static table
try:
    cursor = con.cursor()
    userid = input('Enter the userID in numbners => ')
    name = input('Enter Name => ')
    mobile = input('Enter the Mobile Number => ')
    email = input('Enter the email id => ')
    query = query = f"insert into user(idno, name, mobile, email) values({userid}, '{name.title()}', '{mobile}', '{email.lower()}')"
    cursor.execute(query)
    con.commit()
except Exception as e:
    print(e)

else:
    print()
    print('Record Inserted Sucessfully')
```

```python
# To show the data from tables
tbl_name = input("Enter the name of table")

cursor = con.cursor()
query =f"desc {tbl_name}"
cursor.execute(query)
header =list()
for i in cursor:
    header.append(i[0])

query =f"select * from {tbl_name}"
cursor.execute(query)
data=list()
for i in cursor:
    data.append(i)

for i in range(len(data)):
    for h,v in zip(header,data[i]):
        print(f'{h:<10} : {v}')
    print()
```

```python
# to Delete the record from table
tbl_name= input('Enter the table name from which you want to delete the record')
col_name = input('Enter the column name from which you want to delete the record')
try:
    cursor = con.cursor()
    query = f'desc {tbl_name}'
    cursor.execute(query)
    header=list()
    for i in cursor:
        header.append(i[0])
    if col_name not in header:
        raise Exception (f"{col_name} is not present in the table")
    val = input('Enter the Value of column to delete the record')

    query = f'delete from {tbl_name} where {col_name} = "{val}" '
    cursor.execute(query)
    con.commit()
except Exception as e:
    print(e)
else:
    print('Record Deleted')
```
