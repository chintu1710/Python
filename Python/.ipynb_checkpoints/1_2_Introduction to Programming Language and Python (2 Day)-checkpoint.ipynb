### 8. Dynamically Typed

Python is a dynamically-typed language. That means the type (for example- int, double, long, etc.) for a variable is decided at run time not in advance because of this feature we do not need to specify the type of variable.


### Allocating Memory Dynamically
In Python, the variable data type does not need to be specified. The memory is automatically allocated to a variable at runtime when it is given a value. Developers do not need to write int y = 10 if the integer value 10 is set to y. You may just type y=10.

```python
# in C or Java
# int a = 10      variable a can store int data type only
a = 10 # int
print(type(a), " = ", a)
a = 20.5 # float
print(a)
print(type(a), " = ", a)
a = 'Python' # string
print(a)
print(type(a), " = ", a)
```

```python
a = 10
b = 10
print('value of a = ', a)
print('value of b = ', b)
print('Location of a =', id(a))
print('Location of b =', id(b))
```

```python
a = "Ram"
b = "Ram"
print('value of a = ', a)
print('value of b = ', b)
print('Location of a =', id(a))
print('Location of b =', id(b))
```

## **print()**

```python
# print --> It is a builtin function that print the output on th console
# in print function ---> sep and end parameter are given we can use them as per need
print(print.__doc__)
```

```python
list1 = ['Aishwarya', 'Karishma', 'Kapil', 'Virat', 'Rohit'] # Default value of end = '\n' that means new line
for i in list1:
    print(i)
```

```python
list1 = ['Aishwarya', 'Karishma', 'Kapil', 'Virat', 'Rohit'] #  change value of end = '==='
for i in list1:
    print(i, end = "===")
```

```python
li = ['Aishwarya','Karina', 'Kapil', 'Virat', 'Aditya']
for i in li:
    print(i, end = "------")
```

```python
# sep ---> it seperate the two or more variable from each other defalut value = " "
print('Aishwarya',"Karina", 'Kapil','Virat', 'Aditya') # Default value of sep = ' '
```

```python
# seperate the two variable with +++ or you can specify any character

print('Aishwarya',"Karina", 'Kapil','Virat', 'Aditya', sep = " >>> ")
```

```python
print('Aishwarya',"Karina", sep="$$$$$")
```

```python
print('Aish', 'Kamal', sep="+++")
```
