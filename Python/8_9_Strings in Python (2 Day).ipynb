#### What is string in python?
A string in Python is an immutable sequence of characters. Strings are represented in Python by the str type. Strings can be created by enclosing a sequence of characters in either single or double quotes or triple quotes.
It is a sequence data type. It has indexing.

### What is immutable?
An object is immutable when its state doesn't change after it has been initialized.
For example, String is an immutable class and, once instantiated,the value of a String object never changes.
