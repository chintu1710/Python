````python
playerA = 0
playerB = 0
for i in range(1000):
    a = rd.randint(1,6)
    # print("a", a)
    playerA +=a
    b = rd.randint(1,6)
    # print("b", b)
    playerB +=b

if playerA > playerB:
    print('A Winns  the game')
elif playerA == playerB:
    print('It is a tie')
else:
     print('B Winns  the game')
print(playerA, playerB)
````
