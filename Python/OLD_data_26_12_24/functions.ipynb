# **3 Arbitrary Argument, args '*'.**
# If you do not know how many arguments that will be passed into your function, add a * before the parameter name in the function definition.

# This way the function will receive a **tuple** of arguments, and can access the items accordingly:

```python
def arb_f(*a): # This is parameter # it can take n number of argument in form of tuple
    print(a)
    print(type(a))
    print(f'Score of the First Candidate = {a[0]}')
    print(f'Socre of the Second Candidate = {a[3]}')
    print(f'This is name = {a[2]}')
    return a[0]+a[3]
```

```python
arb_f(88, 97, 'Dilip', 93)
```

```python
sum(2)
2
sum(2,3,1,4)
9
```

```python
def addition(*a):
    res = 0
    for i in a:
        res+=i
    return res
```
