## dict
A dictionary in Python is a collection of key-value pairs. It is unordered,changeable and indexed*.

it is collection of unique keys only

Indexed according to keys

## Method of Dict
1. clear()
2. copy()
3. fromkeys()
4. get()
5. items()
6. keys()
7. pop()
8. popitem()
9. setdefault()
10. update()
11. values()

```python
# Method 1 for Creation of Dict
d = {'a': 'apple', 'm': 'money', 'b': 'book', 'a': 'america', 'b': 'basket'}
print(d)
```

```python
#  use of dict constructor
li = ['a','b','c', 'd']
d2 = dict(li) # error  becasue there is no key value pair 
print(d2)
```

```python
li = ['a','b','c', 'd']
list_of_list = [['a','apple'], ['b', 'book'], ['c', 'cherry']]
d2 = dict(list_of_list) # error  becasue there is no key value pair 
print(d2)
```

```python
for i in li:
    print(i)
```

```python
for i in list_of_list:
    print(i)
```

```python
# Methods to add new key value pair in dict
# Method 1
di = {'m': 'money', 'n':'netflix'}
di['a'] = 'America'
print(di)
```

```python
# Method 2 for add new key value pair
di = {'m': 'money', 'n':'netflix'}
di.update({'z': 'zebra'})
print(di)
di.update([['p','push'],['r', 'rose']])
print(di)
```

```python
# Method 3  to add new key value pair
# setdefault
di = {'m': 'money', 'n':'netflix'}
di.setdefault('m', 'most important') # if key already present then no changes
print(di)
```

```python
di = {'m': 'money', 'n':'netflix'}
di.setdefault('k', 'King') # if key already present then no changes
                           # if not then add new key value pair 
print(di)
```

```python
# Calling of value from dict
# Method 1  direct
di = {'m': 'money', 'n': 'netflix', 'z': 'zebra', 'p': 'push', 'r': 'rose'}
# try to print value of r
print(di['r'])
# try to print value of n
print(di['n'])

# try to print value of key which is not present
print(di['x'])
```

```python
# Calling of value from dict
# Method 2  get
di = {'m': 'money', 'n': 'netflix', 'z': 'zebra', 'p': 'push', 'r': 'rose'}
# try to print value of r
print(di.get('r'))
# try to print value of n
print(di.get('n'))
# try to print value of key which is not present
print(di.get('x'))
```

```python
# special use of get method 
# try to print value of key which is not present if not present then give 
# some message
print(di.get('x','Given key is not presetn in dict'))
```

```python
print(di.get('p','Given key is not presetn in dict'))
```

```python
# From Keys Method
di = {}
a = ['a', 'b', 'c']
v = 1
di = di.fromkeys(a, v)
print(di)
```

```python
di = {'m': 'money', 'n': 'netflix', 'z': 'zebra', 'p': 'push', 'r': 'rose'}
```

```python
# popitem
# it remove the last key value pair from dict
del_value = di.popitem()
print(di)
```

```python
print(del_value)
```

```python
# pop 
# it remove the given key value pair
di = {'m': 'money', 'n': 'netflix', 'z': 'zebra', 'p': 'push', 'r': 'rose'}
del_val = di.pop('n')
print(del_val)
print(di)
```

```python
# clear
# it remove all key value pair from dict
di = {'m': 'money', 'n': 'netflix', 'z': 'zebra', 'p': 'push', 'r': 'rose'}
di.clear()

print(di)
```
