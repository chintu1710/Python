# What is the __init__ method does in class


In Python, the `__init__` method is a special method in a class, often referred to as the **constructor**. It is automatically called when a new instance of the class is created. The main purpose of the `__init__` method is to initialize the attributes of the class and set up the object with any necessary starting values.

### Syntax of `__init__`:
```python
class ClassName:
    def __init__(self, parameters):
        # Initialization code
```

### Key Points:
1. The `__init__` method is not mandatory in a class but is commonly used to define instance attributes.
2. The first parameter of `__init__` is always `self`, which refers to the instance being created. It allows the method to modify the instance's attributes.
3. Additional parameters can be included to initialize the instance with custom data.

### Example:

```python
class Person:
    def __init__(self, name, age):
        self.name = name  # Initialize name attribute
        self.age = age    # Initialize age attribute

# Creating an instance of the Person class
person1 = Person("Alice", 25)

# Accessing the initialized attributes
print(person1.name)  # Output: Alice
print(person1.age)   # Output: 25
```

### When is `__init__` called?
The `__init__` method is automatically invoked when you create a new instance of the class using the class name.

```python
p = Person("John", 30)  # Calls the __init__ method
```

If you don't define an `__init__` method, Python will use a default constructor that does nothing.

### **What Are Dunder (Magic) Methods in Python?**

Dunder (short for "double underscore") or magic methods are special methods in Python that have double underscores at the beginning and end of their names (e.g., `__init__`, `__repr__`, `__add__`). These methods allow you to define the behavior of objects for built-in operations and enable operator overloading, object representation, and custom behavior in your classes.

### **Examples of Common Dunder Methods**
- `__init__(self, ...)`: Constructor to initialize an object.
- `__repr__(self)`: Provides a string representation of the object for developers.
- `__str__(self)`: Provides a user-friendly string representation of the object.
- `__add__(self, other)`: Defines behavior for the `+` operator.
- `__len__(self)`: Defines behavior for the `len()` function.

---

### **`__repr__` Method**
The `__repr__` method is a dunder method that provides a string representation of the object, primarily for debugging purposes. It is called when:
1. You use `repr(object)`.
2. You inspect the object in an interactive session.
3. The object is included in a container like a list or dictionary.

### **Purpose of `__repr__`**
The string returned by `__repr__` should ideally be unambiguous and, if possible, resemble the code needed to recreate the object.

### **Example of `__repr__` Implementation**
```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __repr__(self):
        return f"Person(name='{self.name}', age={self.age})"

# Creating an instance of the Person class
person1 = Person("Alice", 30)

# Using __repr__
print(repr(person1))  # Output: Person(name='Alice', age=30)

# When inspecting the object in an interactive session
person1  # Output: Person(name='Alice', age=30)

# When included in a container
people = [person1, Person("Bob", 25)]
print(people)  # Output: [Person(name='Alice', age=30), Person(name='Bob', age=25)]
```

Encapsulation in Object-Oriented Programming (OOP)

Encapsulation is a fundamental concept of OOP that bundles the data (attributes) and methods (functions) operating on the data into a single unit called a class. It restricts direct access to some components of an object to enforce controlled interaction, ensuring the object's integrity.

In Python, encapsulation is implemented using:


Public Members: Accessible from anywhere.

Protected Members: Indicated with a single underscore _, accessible within the class and its subclasses.
Private Members: Indicated with double underscores __, not directly accessible from outside the class.


Advantages of Encapsulation:

Data Hiding: Restricts access to sensitive data and hides the internal implementation details.

Controlled Access: Provides controlled access to attributes through methods like getters and setters.

Improved Maintainability: Changes to internal workings of a class can be made without affecting external code.

Enhanced Security: Reduces unintended interference or misuse of data.

# Data Hiding
# Encapsulation

````python
class datahide:
    def __init__(self, name, age):
        self._name = name # protected data
        self.__age = age # private data
    def show(self):
        print(f'Name = {self._name}')
        print(f'Age = {self.__age}')
        
````

# type of variable based on class

# 1. class variable
# 2. object variable (object Attribute) / instance variable

# Class Variable:
Definition: A variable that is shared among all instances of a class.

Scope: Defined within the class but outside any methods.

Behavior: A single copy is created and shared across all objects of the class.

Access: Accessed using the class name or an object.


# Instance Variable:

Definition: A variable that is specific to each instance of a class.

Scope: Defined inside a class method, typically within the __init__ method using self.

Behavior: Each object has its own separate copy.

Access: Accessed and modified through the object instance.


```
Aspect	      Class Variable	                 Instance Variable
Scope	      Shared across all objects	         Unique to each object
Storage	      Single copy in memory	         Separate copy for each object
Declaration   Inside the class, outside methods	 Inside methods (usually __init__)
Modification  Affects all objects	         Affects only the specific object
```

````python
import random
class testing:
    variable = 0 # class variable
    score = 0
    def __init__(self, name):
        self.name = name # object variable (instance attribute)/ object attribute
        self.score = testing.score
        testing.variable += 1

    def show_data(self): # method (function defined within the class)
        print(f'Name (Object Variable) = {self.name}')
        print(f'Counter (Class Variable) = {testing.variable}')
        print(f'Current Score (Class Variable) = {self.score}')
        
    def play_event(self):
        num = random.randint(1,6)
        # print(num)
        for i in range(5):
            user_num = int(input('Enter your number to guess :'))
            if num == user_num:
                self.score +=10
                print('Score Updated')
                break
            else:
                print('Try for Next value ')           
````

# Inheritance

````python
# Single Inhertance
class Parent:
    def __init__(self, name):
        self.name = name
    def name_show(self):
        print(f'Name = {self.name}')
    def double(self):
        print(f'Name = {(self.name).upper()}')
        print(f'Name = {(self.name).upper()}')
````

````python
p = Parent('raj kumar')
````

````python
p.double()
````

````python
class child(Parent):
    def __init__(self,name, age):
        super().__init__(name)
        self.age = age
    def show(self):
        print(f'name ={self.name}')
        print(f'age = {self.age}')
````

````python
c1 = child('Mohan', 3) 
````

````python
c1.name_show() # parent method
````

````python
c1.double() # parent method
````

# 5. Multiple Inheritance
One child class inherits from multiple parent classes.

Example: Combining features of a Writer and a Painter

````python
class Parent1:
    def __init__(self, p1):
        self.p1 = p1
    def p1show(self):
        print(f'{(self.p1)*5}')
        
class Parent2:
    def __init__(self, p2):
        self.p2 = p2
    def p2show(self):
        print(f'{(self.p2)*2}')
class child_12(Parent1, Parent2):
    def __init__(self, p1, p2, p3):
        Parent1.__init__(self, p1)
        Parent2.__init__(self, p2)
        self.p3 = p3
    def all_show(self):
        print(f'Value P1 = {self.p1}')
        print(f'Value P2 = {self.p2}')
        print(f'Value P3 = {self.p3}')
````

````python
t12 = child_12('p1_hello', 'p2_hi', 'p3_welcome')
t12.p1show() # parent 1
t12.p2show() # parent 2
t12.all_show() # own 
````

# Multilevel Class
### P1--->C1---->C2

````python
class P1:
    def __init__(self, p1):
        self.p1 = p1
    def p1show(self):
        print(f'{(self.p1)*5}')
        
class C1(P1):
    def __init__(self, p1, c1):
        super().__init__(p1)
        self.c1 = c1
    def c1show(self):
        print(f'{(self.c1)*2}')

class c2(C1):
    def __init__(self, p1, c1, c2):
        super().__init__(p1,c1 )
        self.c2 = c2
        
    def show_all(self):
        print(f'Value P1 = {self.p1}')
        print(f'Value C1 = {self.c1}')
        print(f'Value C2 = {self.c2}')
````

````python
c22 = c2('p1_test', 'c1_hello', 'c2_me')
````

````python
c22.p1show()
````

````python
c22.c1show()
````

````python
c22.show_all()
````

# Abstraction

````python
from abc import ABC, abstractmethod
````

````python
class Base(ABC):
    @abstractmethod
    def show_balance(self):
        pass
    @abstractmethod
    def addtwo(self):
        pass
````

````python
class child(Base):
    def childMethod(self):
        print('This is a child method')
````

````python
c1 = child()
````

````python
class child_abs(Base):
    def childMethod(self):
        print('This is a child abs method')
    def addtwo(self):
        print('Created this method due to abstract method')
    def show_balance(self):
        print('Created this method due to abstract method of show balance')
````

````python
ca1 = child_abs()
````

````python
c2 = child()
````
