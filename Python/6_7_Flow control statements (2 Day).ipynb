```python
#------------------------ Loops ----------------------------
# What are loop in any programing language?
# Loops are the function or control statement in any programming language that can repeat a similar task or
# execute the statement until a specified condition or till completion of iteration.


# ============ in python =============
# For loop is made for iteration purpose mainly
# while loop is made to run until the specified condition is true.
```

```python
#----------- For loop ------------
# in python like other languages ther is no need to initiate, a termination condition and increament or decrement for
# for loop. Because it runs till completion of iteration of iterable if break condition is not specified.
```

```python
# Eg.
st = "SHALINI"
for i in st:
    print(i)
```

```python
fruits  = ["apple", 'mango', 'orange', 'peach', 'strawberry']
for i in fruits:
    print(i)
```

```python
# *********************  break ***********************
# It breaks the loop completely if codition get true
fruits  = ["apple", 'mango', 'orange', 'peach', 'strawberry']
for i in fruits:
    if i == 'orange':
        break
    print(i)

# It will print only - apple  and mango - because when value of i will become orange it breaks the loop(stop Loop)
```

```python
# ~~~~~~~~~~~~~~~~~~~~  continue ~~~~~~~~~~~~~~~~~
# It skips the current execution of loop if codition get true
fruits  = ["apple", 'mango', 'orange', 'peach', 'strawberry']
for i in fruits:
    if i == 'orange':
        continue
    print(i)

# It will print only - all except orange
# because when value of i will become orange it skip the all execution for current iteration.
```

```python
# ~~~~~~~~~~~~~~~~~~~~~~  While Loop ~~~~~~~~~~~~~~~~~~~
# Just like other languages in python also
# for while loop we have intiate a variable, declare a termination condition and increament or decreament of the
# intiated varible to get the termination codition
# sometime we are not defining the termination condition to run the infinte loop
# for this we will declare some flag or while True condition
```

```python
# WAP to print Natural numbers from 1 to 5

n = 1 # intilizatio of variable

# condition = n <6  termination condition

while n < 6: # termination condition n should be less then 6

    print(n) # Statement to execute

    n+=1 # increament of variable to achieve the termination condition
            # if we don't increament the varible condition will true always and full the RAM then it will crash the
            # execution of program
```

```python
# ~~~~~~~~~~~~~~~~~~~~~~ Break ~~~~~~~~~~~~~~~~~~~

n = 1 # intilizatio of variable

# condition = n <6  termination condition

while n < 6: # termination condition n should be less then 6
    if n==3:
        break
    print(n) # Statement to execute

    n+=1
# It will print only 1 and 2 because when n become 3 break condition get true and break the loop
```
