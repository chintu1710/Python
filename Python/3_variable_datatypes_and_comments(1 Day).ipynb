```python
# ----------- Data type ---------
# What is data type?
# Every value has its own characterstics on the basis of that we can differentiate them into different category
# These category are known as data type.
# eg. Number without decimal - integer,
# fractional value -- float
# alphabates -- string

# There are two type of data types
# 1. Primitive Data type - Basic data type - int, float, string, complex, bool
# 2. Non Primitive Data type --> Collection of primitive data types


# ----------------Note------------------
#### There no char data type in python. if you define a single character it will be a string of lenght one
```

```python
# ----------------Note------------------
#### There no char data type in python. if you define a single character it will be a string of length one
st = 'S'
print(type(st))
```

```python
# Primitive Data type
# type() --> it returns the data type of variable
var_int = 25 # --> it is integer type
print(type(var_int))
```

```python
# collection data type -- Collection of primitive data types
list1 = ['a', 'b', 'c'] #  list1 is collection of strings (primitive data type)
print(type(list1))
#### types of collection data types in python

```

```python
# Creation of collection data type
# list
# Method 1 use of [ ]
li = [1,2,3,4,5]
# Method 2 Use of constructor
li2 = list((11,22,33,44))
print('li = ',type(li))
print('li2 =',type(li2))
```

```python
# tuple
# Method 1 use of ( )
tup = (1,2,3,4,5)
# Method 2 Use of constructor
tup2 = tuple((11,22,33,44))
print('tup = ',type(tup))
print('tup2 =',type(tup2))
```

```python
# set
# It is a collection of unique values only
# Method 1 use of { } with atleast one element
set1 = {}
print('set1 -- without any element = ',type(set1))
set2 = {1,2,3,5,1,51,5,61,1,4,1,1,1}
# Method 2 Use of constructor
set3 = set((11,22,33,44, 11, 22))
print('set2 = ',set2)
print('set2 = ',type(set2))
print('set3 = ',set3)
print('set3 =',type(set3))
```

```python
# Dictionary - It has {key: value}
# Map data type (hash data type)
# Method 1
dict1 = {'key': 'Value', 'student': 'Shalini'}
print(dict1)
print(type(dict1))
```

```python
# What is Type Casting?
# Conversion of one data type into another datatype is called type conversion
# There are two type of type casting
# 1. implicit type casting (up casting) --> This kind of type casting done by python interpreter automatically
# There is no need of user interption
```

```python
# Examples
a = 25
b = 36.98
c = a+b
print(type(c))
```

```python
a = 98
b = 13
c = a/b # Float Division / Division
print(c)
print(type(c))
```

```python
# 2. Explicit type casting (down casting / Narrow Casting) --> This kind of type casting done by user explicitly.
# There is need of user interption and most of the time is make data loss mean we loss some value
# In this kind of type casting we mostly do float --> int, string --> int or float

b = 25/2; print(b)

print(int(b)) # Explicit type casting
```
