## Tuple
#### Tuple is a collection of hetrogenous data type which immutable in nature. It has indexing.

## Creation of Tuple

````python
# Method 1
t = ('Apple', 'Mango', 'pineapple')
print(type(t))
print(t)
````

````python
# Method 2
t1 = tuple([2,4,61,45]) # constructor method 
print(type(t1))
print(t1)
````

````python
t1 = ('Shalini') # it is a string
print(t1)
print(type(t1))
````

````python
t1 = ('Shalini',) # if you want to create a tuple with one element only then use ,
print(t1)
print(type(t1))
````

````python
t1 = (23,) # if you want to create a tuple with one element only then use ,
print(t1)
print(type(t1))
````

````python
t1 = tuple('Shalini') 
print(t1)
print(type(t1))
````

## Tuple is immutable

````python
tup = ('Aliya', 'Bhatt')
print(tup)
tup[1] = 'Kapoor'
print(tup)
````

## Methods of tuple
### 1. count()
### 2. index()

````python
# count()
color = ('red', 'green','yellow', 'cherry','yellow', 'brwon', 'white', 'red', 'red')
print(color)
print(color.count('red'))
````

````python
# Index
color = ('red', 'green','yellow', 'cherry','yellow', 'brwon', 'white', 'red', 'red')
print(color)
print(color.index('red'))
````

````python
color = ('red', 'green','yellow', 'cherry','yellow', 'brwon', 'white', 'red', 'red')
print(color)
print(color.index('red',1))
````

````python
color = ('red', 'green','yellow', 'cherry','yellow', 'brwon', 'white', 'red', 'red')
print(color)
print(color.index('red',1,6))
````

## What if We want to modify the tuple
```
Directly it is not possible to do the modification but we can type cast into list and apply the needful modification as then again convert it into tuple

````python
tup = ('Aliya', 'Bhatt')
print(tup)
tup = list(tup) # Type cast into list from tuple
tup[1] = 'Kapoor' # apply modification
tup = tuple(tup) # again type cast into tuple from list
print(tup)
````

## upaking of tuple

````python
t = ('Dilip', 'Sahu', 'Python', 'good','Trainer')
a, b,c,d,e = t
print(a)
print(b)
print(c)
print(d)
print(e)
````

````python
t = ('Dilip', 'Sahu', 'Python', 'good','Trainer')
a, *b = t
print(a)
print(b)
````

````python
t = ('Dilip', 'Sahu', 'Python', 'good','Trainer')
a, *b, c = t
print(a)
print(b)
print(c)
````
